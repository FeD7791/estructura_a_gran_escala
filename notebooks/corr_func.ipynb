```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

Se encuentra que la funcion de correlacion se puede expresar como un power law, concretamente: $$\xi(r) = ({\frac{r}{r_{0}}})^{- \gamma}$$

```python
def xi(r, r0, gamma):
    return  (r/r0)**(-gamma)
```

```python
r = np.arange(10,100,1)
ax = sns.lineplot(x=r, y=xi(r=r,r0=15, gamma=2.0) , label="r0=15")
ax = sns.lineplot(x=r, y=xi(r=r,r0=10, gamma=2.0), label="r0=10")
ax = sns.lineplot(x=r, y=xi(r=r,r0=5, gamma=2.0), label="r0=5")
ax.set_yscale("log")
```
